# Step 8: Exploratory Data Analysis
Univariate, bivariate, multivariate, trend, correlation, and distribution analysis.

```python
import sys
sys.path.append('..')
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')

orders = pd.read_csv('../data/cleaned/orders_features.csv',
                      parse_dates=['order_purchase_timestamp', 'order_delivered_customer_date'])
customer_features = pd.read_csv('../data/cleaned/customer_features.csv')
seller_features = pd.read_csv('../data/cleaned/seller_features.csv')
reviews = pd.read_csv('../data/cleaned/order_reviews.csv')
order_items = pd.read_csv('../data/cleaned/order_items.csv')
```

## Univariate Analysis

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
sns.histplot(orders['total_order_value'].dropna(), bins=50, ax=axes[0])
axes[0].set_title('Distribution of Order Value')
axes[0].set_xlim(0, 1000)

sns.countplot(x='review_score', data=reviews, ax=axes[1])
axes[1].set_title('Review Score Distribution')
plt.tight_layout()
plt.show()
```

## Bivariate Analysis: Delivery Delay vs Review Score

```python
merged = orders.merge(reviews, on='order_id')
delay_summary = merged.groupby('is_delayed')['review_score'].mean().reset_index()

plt.figure(figsize=(6, 4))
sns.barplot(x='is_delayed', y='review_score', data=delay_summary)
plt.xticks([0, 1], ['On-Time', 'Delayed'])
plt.title('Avg Review Score: On-Time vs Delayed')
plt.show()
```

## Trend Analysis: Monthly Revenue

```python
orders['month'] = orders['order_purchase_timestamp'].dt.to_period('M').astype(str)
monthly_revenue = orders.groupby('month')['total_order_value'].sum().reset_index()

plt.figure(figsize=(12, 4))
sns.lineplot(x='month', y='total_order_value', data=monthly_revenue, marker='o')
plt.xticks(rotation=45)
plt.title('Monthly Revenue Trend')
plt.tight_layout()
plt.show()
```

## Correlation Analysis

```python
corr_df = orders[['total_order_value', 'delivery_days', 'delivery_delay_days']].dropna()
plt.figure(figsize=(5, 4))
sns.heatmap(corr_df.corr(), annot=True, cmap='coolwarm', center=0)
plt.title('Correlation: Order Value & Delivery Metrics')
plt.show()
```

## Multivariate Analysis: Customer Spending vs Order Count

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(x='customer_order_count', y='customer_total_spending',
                 hue='is_repeat_customer', data=customer_features.sample(min(5000, len(customer_features))))
plt.title('Customer Order Count vs Total Spending')
plt.show()
```
