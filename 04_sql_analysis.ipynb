# Step 6 & 10: Store Cleaned Data in SQL + SQL Business Analysis
Loads cleaned CSVs into the project database, then runs the business queries
from `streamlit/queries.py` (joins, subqueries, CTEs, window functions, aggregations).

## Load cleaned data into SQL (SQLite by default; MySQL via `.env`)

```python
import sys, os
sys.path.append(os.path.abspath('../streamlit'))
import pandas as pd
from database import get_engine

engine = get_engine()
CLEAN = '../data/cleaned/'

tables = {
    'customers': 'customers.csv',
    'sellers': 'sellers.csv',
    'category_translation': 'category_translation.csv',
    'products': 'products.csv',
    'orders': 'orders.csv',
    'order_items': 'order_items.csv',
    'order_payments': 'order_payments.csv',
    'order_reviews': 'order_reviews.csv',
    'geolocation': 'geolocation.csv',
}

for table_name, csv_file in tables.items():
    df = pd.read_csv(CLEAN + csv_file)
    df.to_sql(table_name, engine, if_exists='replace', index=False)
    print(f'Loaded {table_name}: {len(df)} rows')
```

## Verify load — quick row count check

```python
for table_name in tables:
    n = pd.read_sql(f'SELECT COUNT(*) as n FROM {table_name}', engine).iloc[0]['n']
    print(f'{table_name}: {n} rows in DB')
```
