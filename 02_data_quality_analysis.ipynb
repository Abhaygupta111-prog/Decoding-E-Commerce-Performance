Identify issues in raw data before cleaning: missing values, duplicates, bad types, invalid values, outliers, PK uniqueness.

```python
import sys
sys.path.append('../streamlit')
import pandas as pd
from utils import load_all_csvs, inspect_all, check_pk_uniqueness

dfs = load_all_csvs('../data/raw/')
print('Tables loaded:', list(dfs.keys()))
```

## Missing values, duplicates, dtypes per table

## Primary key uniqueness

```python
check_pk_uniqueness(dfs['olist_orders_dataset'], 'order_id', 'orders')
check_pk_uniqueness(dfs['olist_customers_dataset'], 'customer_id', 'customers')
check_pk_uniqueness(dfs['olist_products_dataset'], 'product_id', 'products')
check_pk_uniqueness(dfs['olist_sellers_dataset'], 'seller_id', 'sellers')
check_pk_uniqueness(dfs['olist_order_reviews_dataset'], 'review_id', 'order_reviews')
```

## Invalid / inconsistent categorical values

```python
print('order_status values:')
print(dfs['olist_orders_dataset']['order_status'].value_counts())
print()
print('payment_type values:')
print(dfs['olist_order_payments_dataset']['payment_type'].value_counts())
```

## Outlier check on price / freight / payment value

```python
order_items = dfs['olist_order_items_dataset']
payments = dfs['olist_order_payments_dataset']

print('price stats:\n', order_items['price'].describe())
print()
print('freight_value stats:\n', order_items['freight_value'].describe())
print()
print('payment_value stats:\n', payments['payment_value'].describe())
```
